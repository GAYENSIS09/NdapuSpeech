```python
import logging
logging.basicConfig(level=logging.INFO, format='%(message)s')
```

```python
from ndapuspeech import config
config.ensure_dirs()
print(config.get_config()['config'])
print(config.get_config()['training'])
```

```python
from ndapuspeech import models

result = models.train_tokenizer_pipeline(
    vocab_size=config.DATA['tokenizer']['vocab_size'],
    extend_whisper=False,   # True pour étendre un tokenizer Whisper
)
result
```

```python
print(f"{result['n_pieces']} pièces BPE apprises sur {result['corpus_chars']} caractères")
if result.get('extended_tokenizer'):
    print('Tokenizer étendu :', result['extended_tokenizer'])
```

# 03 — Construction du dataset unifié train/val/test

Fusionne toutes les sources (téléchargées + manifests YouTube) en un dataset dédupliqué, réparti en **train/val/test** (stratifié par source, ratio depuis `configs/training.yaml`). Les chemins audio sont normalisés en chemins absolus.

```python
import logging
logging.basicConfig(level=logging.INFO, format='%(message)s')
```

```python
from ndapuspeech import config
config.ensure_dirs()
print(config.get_config()['config'])
print(config.get_config()['training'])
```

```python
from ndapuspeech import data, config

summary = data.build_unified_dataset(
    output_dir=config.PROCESSED_DATA_DIR / 'ndapuspeech_ds',
    train_ratio=config.TRAIN_SPLIT,
    push_to_hub=None,   # ex: 'ndapuspeech/ndapuspeech-dataset'
)
summary
```

```python
print(f"Durée totale : {summary['total_hours']:.1f} h — {summary['n_samples']} segments")
print('Heures par source :')
for src, hrs in sorted(summary['by_source'].items(), key=lambda x: -x[1]):
    print(f'  {src:20s} {hrs:6.1f} h')
print('Répartition des splits :', summary['splits'])
```

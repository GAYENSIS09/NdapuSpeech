```python
import logging
logging.basicConfig(level=logging.INFO, format='%(message)s')
```

```python
from ndapuspeech import config
config.ensure_dirs()
print(config.get_config()['config'])
print(config.get_config()['training'])
```

```python
from ndapuspeech import inference, config

audio_files = [
    'data/youtube/segments/sample_1.wav',
    'data/youtube/segments/sample_2.wav',
]
results = inference.transcribe_files(
    model_path='models/ndapuspeech-v1',
    audio_paths=audio_files,
    lm_path=None,             # ou config.DATA['kenlm']['output_bin']
)
results
```

```python
for path, r in results.items():
    print(f'\n--- {path} ---')
    print(r['text'])
```
